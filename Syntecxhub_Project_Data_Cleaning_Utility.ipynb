```python
import pandas as pd

print("Data Cleaning Utility")
```

```python
data = pd.read_csv("/content/data.csv")
```

```python
# Check for missing values

print("Missing values in each column:")
print(data.isnull().sum())
```

```python
# Handle missing Calories values using the median

calories_median = data["Calories"].median()
data["Calories"] = data["Calories"].fillna(calories_median)

print("Calories missing values filled.")
print("Calories median used:", calories_median)
print("Remaining missing Calories:", data["Calories"].isnull().sum())
```

```python
# Check rows around the missing Date value

print(data.loc[20:24, ["Duration", "Date", "Pulse", "Maxpulse", "Calories"]])
```

```python
# Fill the missing Date value

data.loc[data["Date"].isnull(), "Date"] = "2020/12/22"

print("Missing Date value filled.")
print("Remaining missing Date:", data["Date"].isnull().sum())
```

```python
# Check data types

print("Data types before conversion:")
print(data.dtypes)
```

```python
# Convert Date column to datetime

data["Date"] = pd.to_datetime(
    data["Date"].astype(str).str.strip("'"),
    errors="coerce"
)

print("Date column converted to datetime.")
print(data.dtypes)
```

```python
# Check for duplicate rows

duplicate_count = data.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Remove duplicate rows

data = data.drop_duplicates()

print("Duplicate rows removed.")
print("Number of duplicate rows remaining:", data.duplicated().sum())
print("Number of rows after cleaning:", len(data))
```

```python
# Standardize column names

data.columns = (
    data.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("Standardized column names:")
print(data.columns.tolist())
```

```python
# Verify the cleaned dataset

print("Missing values:")
print(data.isnull().sum())

print("\nDuplicate rows:", data.duplicated().sum())

print("\nData types:")
print(data.dtypes)

print("\nCleaned dataset shape:", data.shape)
```

```python
# Find rows where Date could not be converted

print(data[data["date"].isnull()])
```

```python
# Fix the date that could not be parsed

data.loc[26, "date"] = pd.Timestamp("2020-12-26")

print("Row 26 date fixed.")
print(data.loc[26])
```

```python
# Final verification of missing values and duplicates

print("Missing values:")
print(data.isnull().sum())

print("\nDuplicate rows:", data.duplicated().sum())

print("\nDataset shape:", data.shape)
```

```python
# Create a brief cleaning log

cleaning_log = pd.DataFrame({
    "Step": [
        "Missing Calories",
        "Missing Date",
        "Date Parsing",
        "Duplicate Removal",
        "Column Name Standardization"
    ],
    "Action": [
        "Filled missing values with median",
        "Filled missing date using surrounding date sequence",
        "Converted Date column to datetime",
        "Removed duplicate rows",
        "Converted column names to lowercase"
    ],
    "Result": [
        "2 values filled with 291.2",
        "Missing date filled as 2020-12-22",
        "Date type changed to datetime64[ns]",
        "1 duplicate removed",
        "Standardized to lowercase names"
    ]
})

print("Cleaning Log:")
print(cleaning_log)
```

```python
# Save the cleaned dataset

data.to_csv("cleaned_data.csv", index=False)

print("Cleaned dataset saved successfully!")
print("File name: cleaned_data.csv")
```

```python
# Save the cleaning log

cleaning_log.to_csv("cleaning_log.csv", index=False)

print("Cleaning log saved successfully!")
print("File name: cleaning_log.csv")
```
